## Evaluation 

This notebook is responsible for running the TensorFlow Object Detection API evaluation scripts.
The generated evaluation event files are subsequently analysed and interpreted in the
TensorBoard notebook, where validation loss and IoU-based metrics (0.5 and 0.75 thresholds)
are visualised and used to assess model convergence.


### Hardware considerations
Evaluation can be computationally expensive because it performs inference over the evaluation dataset for multiple
checkpoints. To remain mindful of hardware limitations, evaluation is run after training  and relies
on TensorFlow’s built-in logging rather than custom per-image loops.

```python
import os
import time
from pathlib import Path
```

```python
# Set model path
PATH_TO_MODEL = "faster_rcnn_resnet101_v1_1024x1024_coco17_tpu-8"
```

```python
!python model_main_tf2.py \
  --pipeline_config_path=training/TF2/training/{PATH_TO_MODEL}/pipeline.config \
  --model_dir=training/TF2/training/{PATH_TO_MODEL} \
  --checkpoint_dir=training/TF2/training/{PATH_TO_MODEL} \
  --eval_timeout=600 \
  --alsologtostderr
```

## Evaluation Output

The evaluation script was executed in **eval-only** mode using the trained model directory as both `model_dir` and `checkpoint_dir`.  
This writes evaluation summaries (TensorBoard event files) to the `eval/` subdirectory, confirming that evaluation completed successfully.

The resulting detection metrics (including **mAP@0.50IoU** and **mAP@0.75IoU**) are visualised and discussed in the TensorBoard notebook to avoid duplication.

```python
#Checks that event files have been created 
EVAL_DIR = Path("training/TF2/training") / PATH_TO_MODEL / "eval"

event_files = []
if EVAL_DIR.exists():
    event_files = sorted([p for p in EVAL_DIR.glob("events.out.tfevents*")])

print("Eval directory:", EVAL_DIR)
print("Number of eval event files:", len(event_files))

if event_files:
    print("Most recent eval file:", event_files[-1])
    print("Last modified:", os.path.getmtime(event_files[-1]))
```
